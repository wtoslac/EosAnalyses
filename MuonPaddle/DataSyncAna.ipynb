```python
# This block just check if the digitNhits are valid for this dataset.
import ROOT
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors
# Create RDataFrame from the events tree of the input file.
rdf0 = ROOT.RDataFrame("output", "../Data/eos_run_010077.0.root")
timestamp0 = rdf0.AsNumpy(columns=["timestamp"])["timestamp"]
rdf1 = ROOT.RDataFrame("output", "../Data/eos_run_010077.1.root")
timestamp1 = rdf1.AsNumpy(columns=["timestamp"])["timestamp"]
rdf2 = ROOT.RDataFrame("output", "../Data/eos_run_010077.2.root")
timestamp2 = rdf2.AsNumpy(columns=["timestamp"])["timestamp"]


plt.figure(figsize=(16, 10))
evtNum0 = np.arange(0,len(timestamp0))
plt.plot(evtNum0,timestamp0,'r.')
evtNum1 = np.arange(len(timestamp0),len(timestamp0)+len(timestamp1))
plt.plot(evtNum1,timestamp1,'b.')
evtNum2 = np.arange(len(timestamp0)+len(timestamp1),len(timestamp0)+len(timestamp1)+len(timestamp2))
plt.plot(evtNum2,timestamp2,'k.')

plt.ylabel("Timestamp (ns?)", fontsize=18)
plt.xlabel("Event #", fontsize=15)
#plt.ylim(7e13,7e20)
#plt.legend()
plt.show()
```
